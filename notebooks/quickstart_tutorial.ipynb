# Indian Equity Research Toolkit (`indiequity`)
### Quickstart Tutorial: Fundamental Research, Screening & DCF Valuation

This tutorial demonstrates how to use `indiequity` for equity research on Indian markets (NSE & BSE).

```python
import indiequity as ie
print(f"indiequity version: {ie.__version__}")
```

## 1. Fetching Financial Data
Automatically handles Indian tickers and exchange conventions.

```python
profile = ie.fetch_company_data('ITC')
print(f"Company: {profile.company_name} ({profile.symbol})")
print(f"Sector: {profile.sector} | Industry: {profile.industry}")
print(f"Current Price: ₹{profile.current_price:,.2f}")
```

## 2. Computing Key Fundamental Ratios
Calculates ROCE, ROE, D/E, Interest Coverage, and FCF Yield formatted in ₹ Crores.

```python
ratios = ie.calculate_ratios(profile)
print(f"Market Cap:    {ie.format_crores(ratios.market_cap)}")
print(f"ROCE:          {ie.format_percentage(ratios.roce)}")
print(f"ROE:           {ie.format_percentage(ratios.roe)}")
print(f"Debt / Equity: {ratios.debt_to_equity:.2f}")
print(f"Interest Cov:  {ratios.interest_coverage:.1f}x")
print(f"FCF Yield:     {ie.format_percentage(ratios.fcf_yield)}")
```

## 3. DuPont ROE Decomposition
Decomposes Return on Equity into its underlying operational, efficiency, and leverage drivers.

```python
dupont5 = ie.decompose_dupont_5stage(profile)
print(f"Tax Burden:       {dupont5.tax_burden:.2f}x")
print(f"Interest Burden:  {dupont5.interest_burden:.2f}x")
print(f"Operating Margin: {ie.format_percentage(dupont5.operating_margin)}")
print(f"Asset Turnover:   {dupont5.asset_turnover:.2f}x")
print(f"Leverage:         {dupont5.equity_multiplier:.2f}x")
print(f"Resulting ROE:    {ie.format_percentage(dupont5.calculated_roe)}")
print(f"Diagnosis:        {dupont5.interpretation}")
```

## 4. 5-Year FCFF DCF Valuation & Sensitivity Analysis

```python
dcf = ie.calculate_dcf(profile, forecast_years=5, terminal_growth_rate=0.05)
print(f"Intrinsic Value: ₹{dcf.intrinsic_value_per_share:,.2f} per share")
print(f"Current Price:   ₹{dcf.current_price:,.2f}")
print(f"Upside/Downside: {dcf.upside_downside_pct * 100:+.1f}%")

# Display 2D Sensitivity Table
dcf.sensitivity_table
```

## 5. Stock Screening with Preset Criteria

```python
screener = ie.StockScreener(criteria=ie.PRESET_QUALITY_SMALLCAP)
results = screener.screen(['ITC', 'HAPPSTMNDS', 'CAMS', 'RADICO'])
results[['Symbol', 'Market Cap (₹ Cr)', 'ROCE (%)', 'D/E', 'Passed', 'Notes']]
```
